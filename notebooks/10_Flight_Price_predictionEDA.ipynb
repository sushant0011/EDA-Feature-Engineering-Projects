```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df=pd.read_excel('flight_price.xlsx')
df.head()
```

```python
## Feature Engineering process

df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
##Drop 
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
```

```python
df['Arrival_hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_minutes']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(3)
```

```python
df['Arrival_hour'] = df['Arrival_hour'].astype(int)
df['Arrival_minutes']=df['Arrival_minutes'].astype(int)
```

```python
df.info()
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
df['Departure_hour']=df['Dep_Time'].str.split(':').str[0]
df['Departure_minutes']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Departure_hour']=df['Departure_hour'].astype(int)
df['Departure_minutes']=df['Departure_minutes'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].fillna(df['Total_Stops'].mode()[0])
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4})
```

```python
df.isnull().sum()
```

```python
df['Route']=df['Route'].fillna(df['Route'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df.head(2)
```

```python
df['Duration_hours'] = df['Duration'].str.split(' ').str[0].str.split('h').str[0]
df['Duration_minutes']=df['Duration'].str.split('m').str[0].str.split(' ').str[1]
```

```python
df.head(2)
```

```python
df.info()
```

```python
df['Duration_hours'] = df['Duration_hours'].astype(int)
df['Duration_minutes']=df['Duration_minutes'].astype(int)
```
